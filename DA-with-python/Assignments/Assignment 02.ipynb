14)A student will not be allowed to sit in exam if his/her attendence is less than 75%.

Take following input from user

- Number of classes held

- Number of classes attended.

And print

- percentage of class attended

- Is student is allowed to sit in exam or not.


```python
classes_held=int(input("Enter number of classes held"))
classes_attended =int(input("Enter number of classes attended"))
percent_classes_attended = classes_attended / classes_held * 100
if percent_classes_attended < 75:
  print(f"You have {percent_classes_attended:.2f}% attendance.\n You are not allowed to sit in exam.")
else:
  print(f"You have {percent_classes_attended:.2f}% attendance.\n You are allowed to sit in exam")
```

16) Write a program to check if a year is leap year or not.

If a year is divisible by 4 then it is leap year but if the year is century year like 2000, 1900, 2100 then it must be divisible by 400.

```python
year=int(input("Enter year"))
if  year % 400 == 0:
  print("It is a leap year")
elif year % 100:
  print("It is not a leap year")
elif year % 4 == 0:
  print("It is a leap year")
else:
  print("It is not a leap year")
```
