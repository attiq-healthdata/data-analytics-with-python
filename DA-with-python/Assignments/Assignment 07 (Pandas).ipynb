7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python

import pandas as pd
import numpy as np
# Creating a DataFrame having some missing values(NaN)
people = {
    'Name': ['Asad', 'Zafar', 'Rida', 'Saad', 'Eman'],
    'Age': [23, np.nan, 28, 30, np.nan],
    'Salary': [40000, 52000, np.nan, 61000, 45000],
    'City': ['Delhi', 'Lahore', 'Karachi', np.nan, 'Multan']
}

df = pd.DataFrame(people)

print("Missing values:")
print(df.isnull().sum())

# Fill missing numeric values with column means
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

# Drop rows that still contain missing values
df.dropna(inplace=True)

print("\nData after cleaning:")
print(df)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python

# Salary sorted from highest to lowest and age from in ascending order

sorted_df = employee_df.sort_values(
    by=['Salary', 'Age'],
    ascending=[False, True]
)

print(sorted_df)

print("\nTop 3 highest-paid employees:")
print(employee_df.nlargest(3, 'Salary'))
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python

city_data = employee_df.groupby('Location').agg({
    'Salary': ['mean', 'max'],
    'Name': 'count'
})

city_data.rename(
    columns={'count': 'Employee_Count'},
    inplace=True
)

print(city_data)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python

import pandas as pd

employees = pd.DataFrame({
    'EmpID': [1, 2, 3, 4],
    'Name': ['Zaid', 'Bilal', 'Salman', 'Usman'],
    'DeptID': [101, 102, 101, 103]
})

departments = pd.DataFrame({
    'DeptID': [101, 102, 104],
    'Department': ['IT', 'HR', 'Finance']
})

# Inner join
inner_data = pd.merge(employees, departments, on='DeptID', how='inner')
print("Inner join:")
print(inner_data)

# Left join
left_data = pd.merge(employees, departments, on='DeptID', how='left')
print("\nLeft join:")
print(left_data)

# Concatenate two DataFrames
first_part = employees.iloc[:2]
second_part = employees.iloc[1:]

combined = pd.concat([first_part, second_part], ignore_index=True)
combined = combined.drop_duplicates()

print("\nAfter concatenation and removing duplicates:")
print(combined)
```
