Q3 — Loop Through Dictionary
person = {
"name": "Ali",
"age": 25,
"city": "Karachi"
}
Print keys and values using loop

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

for key, value in person.items():
    print(key, ":", value)
```

Q4 — Count Frequency of Numbers
numbers = [1,2,2,3,3,3,4,4,4,4]
Create frequency dictionary.

```python
from collections import Counter
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
frequency = dict(Counter(numbers))

print(frequency)
```

Q5 — Word Counter
Take sentence input and count word frequency

```python
from collections import Counter
sentence = input("Enter a sentence: ").lower()
words = sentence.split()
word_frequency = dict(Counter(words))

print(word_frequency)
```

Q6 — Student Marks & Average
marks = {"Math":80, "English":70, "Physics":90}
Calculate total and average.

```python
marks = {"Math": 80, "English": 70, "Physics": 90}

total_marks = sum(marks.values())
average_marks = total_marks / len(marks)

print(f"Total Marks: {total_marks}")
print(f"Average Marks: {average_marks:.2f}")
```

Q7 — Highest Scoring Student
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}
Find highest scoring student

```python
students = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zara": 95}

highest_student = max(students, key=students.get)

print(f"Highest Scoring Student: {highest_student} ({students[highest_student]} marks)")
```

Q8 — Merge Dictionaries
dict1 = {"a":1,"b":2}
dict2 = {"c":3,"d":4}
Merge and print result

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged_dict = dict1 | dict2

print(merged_dict)
```
