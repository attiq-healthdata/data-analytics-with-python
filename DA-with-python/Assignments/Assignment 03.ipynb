3. Find and display the largest number of a list without using built-in function max(). Your program
should ask the user to input values in list from keyboard

```python
user_input=input("Enter numbers separated by spaces:").split()
nums=[int(x) for x in user_input]
largest=nums[0]
for num in nums:
    if num>largest:
      largest=num
print(largest)
```

4. Write a program that rotates the element of a list so that the element at the first index moves to the
second index, the element in the second index moves to the third index, etc., and the element in the last
index moves to the first index. 

```python
user_input = input("Enter list elements separated by spaces: ")
items = user_input.split()
if len(items) > 0:
  rotated_list = items[-1:] + items[:-1]
  print( rotated_list)
else:
  print("The list is empty")
```

 6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It
should print the date in the form March 12, 2021. 

```python

date = input("Enter date (mm/dd/yyyy): ")
month, day, year = date.split("/")
months = ["January", "February", "March", "April","May", "June", "July", "August","September", "October", "November", "December"]
month_name = months[int(month) - 1]
print(month_name, day + ",", year)
```

7. Write a program with a function that accepts a string from keyboard and create a new string after
converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses."
the output should be "Stop And Smell The Roses"

```python
def capitalize_words(sentence):
    return sentence.title()
sentence = input("Enter a sentence: ")
new_string = capitalize_words(sentence)
print(new_string)
```

8. Find the sum of each row of matrix of size m x n

```python
# Take number of rows and columns
m = int(input("Enter number of rows: "))
n = int(input("Enter number of columns: "))

# Create an empty matrix
matrix = []

# Input the matrix
for i in range(m):
    row = []
    for j in range(n):
        value = int(input("Enter value: "))
        row.append(value)
    matrix.append(row)

# Find the sum of each row
for i in range(m):
    total = 0

    for j in range(n):
        total = total + matrix[i][j]

    print("Sum of row", i + 1, "=", total)
```

9. Write a program to add two matrices of size n x m

```python
n = int(input("Enter number of rows: "))
m = int(input("Enter number of columns: "))

A = []

for i in range(n):
    row = list(map(int, input(f"Enter row {i + 1} of first matrix: ").split()))
    A.append(row)

B = []

for i in range(n):
    row = list(map(int, input(f"Enter row {i + 1} of second matrix: ").split()))
    B.append(row)

C = []

for i in range(n):
    row = []
    for j in range(m):
        row.append(A[i][j] + B[i][j])
    C.append(row)

print("Sum of the two matrices:")

for row in C:
    print(row)
```

```python
r1 = int(input("Enter number of rows of first matrix: "))
c1 = int(input("Enter number of columns of first matrix: "))

A = []

for i in range(r1):
    row = list(map(int, input(f"Enter row {i + 1} of first matrix: ").split()))
    A.append(row)

r2 = int(input("Enter number of rows of second matrix: "))
c2 = int(input("Enter number of columns of second matrix: "))

B = []

for i in range(r2):
    row = list(map(int, input(f"Enter row {i + 1} of second matrix: ").split()))
    B.append(row)

if c1 != r2:
    print("Matrix multiplication is not possible.")
else:
    C = []

    for i in range(r1):
        row = []
        for j in range(c2):
            total = 0
            for k in range(c1):
                total = total + A[i][k] * B[k][j]
            row.append(total)
        C.append(row)

    print("Product of the two matrices:")

    for row in C:
        print(row)
```
