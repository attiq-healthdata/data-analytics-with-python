QUESTION 1: Working with Lists
1.	Create a list called numbers containing at least 8 integers.
2.	Using a for loop, do the following:
o	Print all numbers in the list.
o	Print the sum of all numbers.
o	Print only the even numbers.
Requirements:
•	Use at least one loop.
•	Use conditional statements.
•	Display clear output.


```python
#Creating a list "numbers" containing 8 integers
numbers = [5,15,27,8,30,16,34,4]

total = 0

# using for loop, printing all numbers in the list
for number in numbers:
    print(number)
    total = total + number 
print("Even numbers:", end=" ")
 # under for loop using conditional statement,printing only the even number
for number in numbers:
    if number % 2 == 0:
        print(number , end=" ")
 # printing the sum of all numbers       
print("\nSum =",total)  

```

QUESTION 2: Dictionary Basics
1.	Create a dictionary called student with the following keys:
o	name
o	age
o	course
o	marks (must store a list of 3 marks)

```python
# A1. Creating a dictionary called student with the required keys

student = {
"name": "Sarah",
"age": 20,
"course": "Computer Science",
"marks": [70,80,90]}

# printing all keys and their values
print("Student Information")

for key, value in student.items():
  print(f"{key.capitalize()}: {value}")

# calculating the average of the marks

total_marks = 0
marks_list = student["marks"]

for mark in marks_list:
  total_marks += mark

average = total_marks / len(marks_list)
print(f"\nAverage Marks: {average:.2f}")

# Displaying whether the student Passed or Failed 

if average >= 50:
  print("Result: Passed")
else:
  print("Result: Failed")
```

QUESTION 3: List of Dictionaries
1.	Create a list called employees.
2.	Each employee must be stored as a dictionary containing:
o	name
o	department
o	salary
3.	Using loops:
o	Print details of all employees.
o	Find the employee with the highest salary.
o	Calculate the total salary expense.
Requirements:
•	Minimum 3 employees.
•	Use comparison logic.
•	Use loops to process data.

```python
# Create a list of employees
employees = [
    {"name": "Asad", "department": "Science", "salary": 80000},
    {"name": "Zaid", "department": "Commerce", "salary": 60000},
    {"name": "Ahmed", "department": "HR", "salary":50000}
]

total_salary = 0
highest_salary = 0
highest_employee = ""

# Printing details of all employees
for employee in employees:

    print("Name:", employee["name"])
    print("Department:", employee["department"])
    print("Salary:", employee["salary"])
    print()

    total_salary = total_salary + employee["salary"]

    if employee["salary"] > highest_salary:
        highest_salary = employee["salary"]
        highest_employee = employee["name"]

print("Highest Salary:", highest_salary)
print("Employee with Highest Salary:", highest_employee)

print("Total Salary Expense:", total_salary)
```

QUESTION 4: While Loop with User Input
1.	Ask the user to enter numbers.
2.	Store the numbers in a list.
3.	Stop when the user enters -1.
4.	After the loop ends:
o	Print the complete list.
o	Print the largest number.
o	Print the smallest number.
Requirements:
•	Must use a while loop.
Must store input in a list

```python
### Initialize an empty list to store the inputs

user_numbers =[] 

# Must use a while loop

while True:
  user_input = int(input("Enter a number (-1 to stop):"))

### Stop when the user enters -1

  if user_input == -1:
    break

### Store the numbers in a list

  user_numbers.append(user_input)

print("Numbers entered:",user_numbers)

### After the loop ends, process and display the results if the list is not empty

if len(user_numbers) > 0:
  
  print("Largest number:",max(user_numbers))
  print("Smallest Number:",min(user_numbers))
else:
  print("No numbers were entered.")
```

QUESTION 5: Word Frequency Counter
1.	Ask the user to enter a sentence.
2.	Convert the sentence into a list of words.
3.	Use a dictionary to count how many times each word appears.
4.	Print the word frequency clearly.
Example Format (for understanding only):
apple: 3
banana: 2
orange: 1
Requirements:
•	Use string splitting.
•	Use a dictionary for counting.
•	Use loops for processing


```python
# A.5 Asking the user to enter a sentence.
User_input = input("Enter a sentence")

# Converting a sentence into a list of words by using string splitting
words = User_input.lower().split()

# Using an empty dictionary to count how many times each word appeares
word_count = {}

# using a loop for processing 
for word in words:
  # if statements is used to add 1 to word, if it is already in dictionary
  if word in word_count:
    word_count[word] = word_count[word] + 1

 # if it is new word, add it within dictionary with a count of one
  else:
    word_count[word] = 1

print("Word Frequency:")

for word, count in word_count.items():
  print(word,":",count)
  

```

QUESTION 6: Nested Loops – Multiplication Table
Using nested loops, print a multiplication table from 1 to 5.
Expected Output Format:
1 2 3 4 5
2 4 6 8 10
3 6 9 12 15
4 8 12 16 20
5 10 15 20 25
Requirements:
•	Must use nested loops.
•	Proper formatting required.


```python
# A6.  Multiplication table from 1 to 5, with each table alligned vertically

for row in range(1,6):
    for col in range(1,6):
      print(row*col,end=" ")
    print()
```
